## Creación del nuevo dataset a partir de funciones

#### Notas de 0 a 10 (0 a 1)

```python
def generate_score_0_10(n_records):
    data = {
        'Hours_Studied': np.random.randint(0, 3, n_records),
        'Attendance': np.random.randint(0, 30, n_records),
        'Previous_Scores': np.random.randint(0, 20, n_records),
        'Tutoring_Sessions': np.random.randint(0, 2, n_records),
        'PI_High': np.random.choice([0, 1], n_records, p=[0.9, 0.1]),
        'PI_Low': np.random.choice([0, 1], n_records, p=[0.7, 0.3]),
        'AtR_High': np.random.choice([0, 1], n_records, p=[0.8, 0.2]),
        'AtR_Low': np.random.choice([0, 1], n_records, p=[0.6, 0.4]),
        'Exam_Score': np.random.randint(0, 11, n_records)
    }
    data['PI_Low'] = np.where((data['PI_High'] == 1) & (data['PI_Low'] == 1), 0, data['PI_Low'])
    data['AtR_Low'] = np.where((data['AtR_High'] == 1) & (data['AtR_Low'] == 1), 0, data['AtR_Low'])
    return pd.DataFrame(data)
```
